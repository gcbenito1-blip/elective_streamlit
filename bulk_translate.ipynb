```python
import pandas as pd
import asyncio
from googletrans import Translator
import unicodedata
from tqdm.notebook import trange, tqdm
from time import sleep
import re
```

```python
tgl_df = pd.read_csv('tgl_review.csv', index_col=0)
items = tgl_df['clean'].tolist()
translated = []
```

```python
def translate_bulk_sync(items, batch_size=20):
    results = []
    translator = Translator()

    with tqdm(total=len(items)) as pbar:
        for i in range(0, len(items), batch_size):
            batch = items[i:i+batch_size]

            translations = translator.translate(batch, dest='en')

            results.extend([t.text for t in translations])
            pbar.update(len(batch))

    return results

translated = translate_bulk_sync(items)
tgl_df['translated'] = translated
```
